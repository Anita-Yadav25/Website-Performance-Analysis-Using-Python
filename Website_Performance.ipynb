```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df= pd.read_csv("Website_performance.csv")
```

```python
df.head()
```

```python
df.columns= df.iloc[0]
df.head()
```

```python
df= df.drop(index=0).reset_index(drop=True)
df.columns= ["Primary Channel","DateHour","Users","Sessions","Engaged sessions","Average engagement time per session","Engaged sessions per user","Events per session","Engagement rate","Event count"]
```

```python
df.head()
```

```python
df.info()
```

```python
df["DateHour"]= pd.to_datetime(df["DateHour"],format= "%Y%m%d%H", errors='coerce')
df.head()
```

```python
numeric_cols= df.columns.drop(["Primary Channel","DateHour"])
df[numeric_cols]= df[numeric_cols].apply(pd.to_numeric,errors='coerce')
df["hour"]= df["DateHour"].dt.hour
```

```python
df.head()
```

```python
df.info()
```

```python
df.describe()
```

# ANALYSIS

#Sessions And USERS OVER TIME


```python
 #  What patterns or trends can you observe
 #  in website sessions and users over time?
```

```python
sns.set(style="whitegrid")
```

```python
plt.figure(figsize=(10,5))
df.groupby("DateHour")[["Sessions", "Users"]].sum().plot(ax=plt.gca())
plt.title("Sessions and Users Over Time")
plt.xlabel("DateHour")
plt.ylabel("Count")
plt.show()
```

# Total users by channel

```python
sns.set(style="whitegrid")
plt.figure(figsize=(10,5))
sns.barplot(data=df, x="Primary Channel", y="Users", estimator=np.sum, palette="viridis")
plt.title("Total users by channel")
plt.xticks(rotation=45)
plt.show()
```

# Average engaement time by Channel

```python
sns.set(style="whitegrid")
plt.figure(figsize=(8,5))
sns.barplot(data=df, x="Primary Channel", y="Average engagement time per session", estimator=np.mean, palette="magma")
plt.title("Avg engagement Time by channel")
plt.xticks(rotation=45)
plt.show()
```

# Engagement rate distribution by channels

```python
 sns.set(style="whitegrid")
plt.figure(figsize=(8,5))
sns.boxplot(data=df, x="Primary Channel", y="Engagement rate", palette="coolwarm")
plt.title("Engagement Rate Distribution by Channel")
plt.xticks(rotation=45)
plt.show()
```

# Engaged vs Non Engaged Sessions

```python
df.head()
```

```python
session_df=df.groupby("Primary Channel")[["Sessions", "Engaged sessions"]].sum().reset_index()
session_df["Non-Engaged"]= session_df["Sessions"]- session_df["Engaged sessions"]
session_df_melted= session_df.melt (id_vars="Primary Channel", value_vars=["Engaged sessions", "Non-Engaged"])

sns.set(style="whitegrid")
plt.figure(figsize=(8,5))
sns.barplot(data=session_df_melted, x="Primary Channel", y="value", hue="variable")
plt.title("Engaged Vs Non-Engaged Sessions")
plt.xticks(rotation=45)
plt.show()
```

# Traffic by hour and channel

```python
heatmap_data= df.groupby(["hour", "Primary Channel"])["Sessions"].sum().unstack().fillna(0)

plt.figure(figsize=(12,6))
sns.heatmap(heatmap_data, cmap="YlGnBu", linewidths=.5, annot= True, fmt='0f')
plt.title("Traffic by Hour and Channel")
plt.xlabel("Channel Group")
plt.ylabel("Hour of Day")
plt.show()
```

# Engagement rate vs Sessions Over Time

```python
df.head()
```

```python
sns.set(style="whitegrid")
df_plot= df.groupby("DateHour")[["Engagement rate", "Sessions"]].mean().reset_index()
plt.figure(figsize=(10,5))
plt.plot(df_plot["DateHour"], df_plot["Engagement rate"], label="Engagement rate", color="green")
plt.plot(df_plot["DateHour"], df_plot["Sessions"], label="Sessions", color="blue")
plt.title("Engagement rate vs Sessions Over Time")
plt.xlabel("DateHour")
plt.legend()
plt.show()
```
